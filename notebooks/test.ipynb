```python
import torch
from torch import nn

# giả lập config
class Config:
    def __init__(self):
        self.head = 8
        self.d_model = 256
        self.d_kv = 32
        self.d_ff = 512

# import model của bạn
from models.modules.ocr_encoder import OCREncoder

# tạo config
config = Config()

# tạo embedding (nếu dùng)
vocab_size = 1000
word_embed = nn.Embedding(vocab_size, config.d_model)

# khởi tạo model
encoder = OCREncoder(config, word_embed)

# ===== dummy input =====
batch_size = 2
seq_len = 10

inputs = torch.randint(0, vocab_size, (batch_size, seq_len))  # token ids
ocr_mask = inputs.ne(0).unsqueeze(1).unsqueeze(1)
print(f"ocr_mask: {ocr_mask.shape}")
# ===== forward =====
x, break_probs = encoder(inputs, ocr_mask)

print("Output x shape:", x.shape)
print("Break probs shape:", break_probs.shape)
```
